```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

BASE = Path("..").resolve()
CSV = BASE / "data" / "dataset_simulado.csv"

df = pd.read_csv(CSV)
df["data_hora"] = pd.to_datetime(df["data_hora"])
print(f"Registros: {len(df)}")
df.head()
```

```python
df.info()
df.describe().round(2)
```

```python
print("Distribuição risco_label:")
print(df["risco_label"].value_counts(normalize=True).round(3))
print("\nPor tipo de operação:")
print(pd.crosstab(df["tipo_operacao"], df["risco_label"], normalize="index").round(3))
```

```python
numeric = df.select_dtypes(include="number").drop(columns=["risco_label"], errors="ignore")
corr = numeric.corrwith(df["risco_label"]).sort_values(ascending=False)
print("Correlação com risco_label:")
print(corr.round(3))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.boxplot(data=df, x="risco_label", y="distancia_corpo_dagua", ax=axes[0])
axes[0].set_title("Distância da água vs risco")
axes[0].set_xlabel("Risco (0=baixo, 1=alto)")

sns.boxplot(data=df, x="risco_label", y="umidade_solo", ax=axes[1])
axes[1].set_title("Umidade do solo vs risco")
axes[1].set_xlabel("Risco (0=baixo, 1=alto)")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.heatmap(
    df[numeric.columns.tolist() + ["risco_label"]].corr(),
    annot=True, fmt=".2f", cmap="RdBu_r", center=0, ax=ax
)
ax.set_title("Matriz de correlação — variáveis numéricas")
plt.tight_layout()
plt.show()
```

## Conclusões preliminares

1. **Proximidade da água** e **umidade do solo** são os fatores mais correlacionados com risco alto.
2. Operações em **gleissolo** concentram mais casos de risco (solo encharcado + baixa distância).
3. **Transporte** eleva risco relativo vs operações de campo.
